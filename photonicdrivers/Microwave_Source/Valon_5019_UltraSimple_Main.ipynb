```python
import serial 

connection = serial.Serial(port="COM3", baudrate=115200, timeout=1)

```

```python
connection.isOpen()
```

```python
connection.write(b'ID?\r')
response = connection.readline()
print(response.decode())
```

```python
connection.write(b'ID?\r')
response = connection.readline().decode().strip()
print(response)
```

```python
import time
connection.write(b'*ID?\r\n')
time.sleep(0.5)
print(connection.in_waiting)
print(connection.read(connection.in_waiting))
```

```python
connection.write(b'ID?\r')
time.sleep(0.5)
print(connection.read_all())
connection.write(b'ID?\n')
time.sleep(0.5)
print(connection.read_all())
connection.write(b'ID?\r\n')
time.sleep(0.5)
print(connection.read_all())
connection.write(b'ID?')
time.sleep(0.5)
print(connection.read_all())
```
