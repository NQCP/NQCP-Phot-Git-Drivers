```python
import numpy as np
from labserver.Server.DriverFinder import DriverFinder
from photonicdrivers.TimeTaggers.Swabian_TimeTagger_Driver import Swabian_TimeTagger_Driver
import TimeTagger
finder = DriverFinder()
server_tagger = finder.get_driver("time_tagger")
```

```python
server_ip = "10.209.67.193" # Example IP address
server_port = "41101"  # Example port number
tagger = TimeTagger.createTimeTaggerNetwork(server_ip + ":" + server_port)
```

```python
delayed_channel = TimeTagger.DelayedChannel(tagger=tagger, input_channel=12, delay=1_000)

delayed_channel_number = delayed_channel.getChannel()
print(delayed_channel_number)
histogram = TimeTagger.Histogram(tagger=tagger, start_channel=11,click_channel=delayed_channel_number, binwidth=10_000, n_bins=100)
```

```python
server_delayed_channel_number = server_tagger.get_delayed_channel_number(channel=11, delay_ps=1_000)
print(server_delayed_channel_number)
histogram = TimeTagger.Histogram(tagger=tagger, start_channel=server_delayed_channel_number, click_channel=11, binwidth=10_000, n_bins=100)
```
